# Multiple lines

```python

months =['January', 'feb','March','April','May','June','August','sep','OCtober','Nov','Dec']

product_a= [1000,1200,15000,1700,1600,1800,1500,2000,2200,2500,2800]
product_b = [1500,200,2500,2600,2700,2900,3000,3100,3200,3600,3800]

plt.figure(figsize=(12,8))
plt.plot(months, product_a, marker = 's', linestyle = '--', color = 'g', label = 'Product A')
plt.plot(months, product_b, marker = 'o', linestyle = '-', color = 'r', label = 'Product B')
plt.title('Monthly Sales Data for product A and product B')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

```python
import numpy as np

# generate 500 random std marks (normally distributed around 72)

np.random.seed(42)
marks = np.random.normal(loc =72, scale=12, size = 500).clip(40,100)

# define 10-point bin boundaries ;[40,50,60,70,80,90,100]
bin_edges = range(40,105,10)

plt.figure(figsize=(9,5))

plt.hist(marks, bins = bin_edges, color = 'lightblue', edgecolor='black')
plt.title('Final exam score distribution (N = 500)')
plt.xlabel('Score range')
plt.ylabel('no of stds')
plt.xticks(bin_edges)
plt.grid(axis = 'y',linestyle = '--')
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# generate 500 random observations

np.random.seed(42)
income = np.random.lognormal(mean=8, sigma=0.5, size = 500)

# define 10-point bin boundaries ;[40,50,60,70,80,90,100]
bin_edges = np.linspace(0,15000,31)

plt.figure(figsize=(9,5))

plt.hist(income, bins = bin_edges, color = 'lightblue', edgecolor='black')
plt.title('log normal score distribution (N = 500)')
plt.xlabel('monthly income')
plt.ylabel('frequency')
plt.xticks(bin_edges)
plt.grid(axis = 'y',linestyle = '--')
plt.show()
```

# Piechart

```python
subject=['maths','science','social','neplai','arts']
hours_studied=[1,2,3,4,5]

plt.Figure(figsize=(8,8))
plt.pie(hours_studied,labels=subject,autopct='%1.1f%%',startangle=90, colors=['red','green','yellow','blue','brown'])
plt.title("hours studies by subject")
plt.show()
```

```python
subject=['maths','science','social','neplai','arts']
hours_studied=[1,2,3,4,5]

explode=[0.1,0,0.1,0,0]

plt.Figure(figsize=(8,8))
plt.pie(hours_studied,labels=subject,autopct='%1.1f%%',startangle=90, colors=['red','green','yellow','blue','brown'], explode = explode)
plt.title("hours studies by subject")
plt.show()
```

# Figure and axes

```python
fig,axes = plt.subplots(2,2)

x=[1,2,3,4,5]
y=[2,3,4,5,6] 

axes[0,0].plot(x,y)
axes[0,0].set_title('line')

axes[0,1].bar(x,y)
axes[0,1].set_title('bar')

axes[1,0].scatter(range(5),y)
axes[1,0].set_title('scatter')

axes[1,1].hist(y)
axes[1,1].set_title('histogram')

plt.tight_layout()
plt.show()






```

#  Save Figure

```python
fig,axes = plt.subplots(2,2)
x=[1,2,3,4,5]
y=[2,3,4,5,6]

axes[0,0].plot(x,y)
axes[0,0].set_title('line')

axes[0,1].bar(x,y)
axes[0,1].set_title('bar')

axes[1,0].scatter(range(5),y)
axes[1,0].set_title('scatter')

axes[1,1].hist(y)
axes[1,1].set_title('histogram')


plt.tight_layout()

plt.savefig("my_charts.png",dpi=300,bbox_inches="tight")
plt.show()
```
